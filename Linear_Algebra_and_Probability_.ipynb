**Linear Algebra and Probability**

```python
#Importing essential libraries
import numpy as np
import pandas as pd
```

```python
#Loading the dataset
df = pd.read_csv("winequality-red.csv",sep = ";")
print("Rows and columns : ",df.shape)
print("Columns : ",list(df.columns))
print(df.head())
print(df.isnull().sum())
df = df.fillna(df.mean(numeric_only=True))
```

```python
#Extract alcohol and citric acid as vectors.
alcohol_vector = df["alcohol"].to_numpy()
citric_acid_vector = df["citric acid"].to_numpy()
print("Alcohol vector shape:", alcohol_vector.shape)
print("\nCitric acid vector shape:", citric_acid_vector.shape)
```

```python
#Select two features and compute the covariance matrix.
X= df[["alcohol","density"]].to_numpy()
cov_matrix = np.cov(X.T)
print("Feature matrix X shape:", X.shape)
print("Covariance matrix : ",cov_matrix)
```

```python
#Eigen decomposition on the covariance matrix.
eigen_values, eigen_vectors = np.linalg.eig(cov_matrix)

print("Top 2 Eigenvalues:")
print(eigen_values)

print("\nCorresponding Eigenvectors (each column is an eigenvector):")
print(eigen_vectors)

print("\nEigenvector 1 (for 1.13564827):", eigen_vectors[:, 0])
print("Eigenvector 2 (for 2.68507580e-06):", eigen_vectors[:, 1])
```

```python
#Most common wine quality and distribution
quality_counts = df["quality"].value_counts().sort_index()
most_common = df["quality"].mode()[0]

print("Wine quality counts:\n", quality_counts)
print("\nMost common wine quality score:", most_common)
```

 Quality score 5 is most common (681, 42.6%), followed closely by 6 (638, 39.9%). Together, 5 and 6 make up to 82.5% of the dataset, forming a bell-shaped middle peak with rare low (3, 4) and high (7, 8) scores in the tails.
